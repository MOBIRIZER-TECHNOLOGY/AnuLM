```python
#@title 2. Pick a checkpoint and download it (0.8 GB)
checkpoint = "toonist/AnuLM-Coder-400M"  #@param ["toonist/AnuLM-Coder-400M", "toonist/AnuLM-Translate-400M", "toonist/AnuLM-Hindi-QA-400M", "toonist/AnuLM-Base-400M"]
from huggingface_hub import snapshot_download
local = snapshot_download(checkpoint)
print('downloaded to', local)
```

```python
#@title 3. Launch the demo (a public link appears below; keep this cell running)
import torch
from serve import Engine
from app import build
engine = Engine(local, 'cuda' if torch.cuda.is_available() else 'cpu')
build(engine).launch(share=True)
```
